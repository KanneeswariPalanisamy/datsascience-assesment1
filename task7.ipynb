```python
seven = pd.get_dummies(seven,columns=["State_Name"],dtype = int)
seven.head()
```

```python
staticts = seven.select_dtypes(include = "number")
result = pd.DataFrame({
    "Mean": staticts.mean(),
    "Median": staticts.median(),
    "Mode": staticts.mode().iloc[0]
})
print(result.round(3))
```

```python
result2 = pd.DataFrame({
    "Variance":staticts.var(),
    "Standard variance":staticts.std()
})
print(result2)
```

```python
result3 = pd.DataFrame({
    "Minimum": staticts.min(),
    "Maximum": staticts.max(),
    "Range":staticts.max() - staticts.min()
})
print(result3.astype(int))
```

```python
result4 = pd.DataFrame({
    "Q1(25%)" :staticts.quantile(0.25),
    "Q2(Median)":staticts.quantile(0.50),
    "Q3(75%)":staticts.quantile(0.75),
})
result4["IQR"] = result4["Q3(75%)"] - result4["Q1(25%)"]
print(result4)
```

```python
result5 = pd.DataFrame({
    "Skewness": staticts.skew()
})
print(result5.round(3))
```

```python

column = staticts.columns[0]
sns.histplot(staticts[column],kde="true")
plt.title(f"skewness Distribution of {column}")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()
print("Skewness:",staticts[column].skew())
```

```python
column = staticts.columns[5]
sns.histplot(staticts[column],kde="true")
plt.title(f"skewness Distribution of {column}")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()
print("Kurtosis:",staticts[column].kurt())
```

```python
#Qutilers in IQR Method
```

```python
Q1 =  staticts.quantile(0.25)
Q3 =  staticts.quantile(0.75)
IQR = Q3 - Q1
Lower_limit = Q1 - 1.5 * IQR
Upper_limit = Q3 + 1.5 * IQR

outliers = (
    ( staticts < Lower_limit ) |
    ( staticts > Upper_limit) 
)

print("Lower Bound:\n" ,Lower_limit )
print("Upper Bound:\n" ,Upper_limit )
print("Qutliers in each column:\n" ,  staticts[outliers].count())
```
